Q16. Compare `r`, `w` and `a` Create one program demonstrating: ```text r w a ``` Write a short explanation of what each mode does.

````python
f = open("fn.txt","w")
print(f.write("Python is a "))
f.close()

f = open("fn.txt","a")
print(f.write("Programming Language"))
f.close()

f = open("fn.txt","r")
print(f.read())
f.close()

# 'w' (Write Mode): Opens a file for writing. It overwrites existing content or creates a new file if it doesn't exist.
# 'a' (Append Mode): Opens a file for adding new data at the end without overwriting existing content. Creates a new file if it doesn't exist.
# 'r' (Read Mode): Opens a file for reading its existing content. It does not modify the file.
````
